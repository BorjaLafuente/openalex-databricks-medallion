# Setup
Creates the schema and the volumes used by the pipeline. Run it once, before the other notebooks.

After running it, upload the input files through **Catalog → workspace → openalex → Volumes → Upload to this volume**:

| Volume | Files | Source |
|---|---|---|
| `raw` | `UDC.jsonl`, `USC.jsonl`, `UVIGO.jsonl` | `data/raw/` of [openalex-galician-universities](https://github.com/BorjaLafuente/openalex-galician-universities), created by `src/extract.py` |
| `reference` | `results/*.csv` (six files) | `results/` of the same repository, used by `04_validation` |

```sql
-- All objects are idempotent: the notebook can be re-run safely
CREATE SCHEMA IF NOT EXISTS workspace.openalex
COMMENT 'OpenAlex publications of the Galician universities (UDC, USC, UVIGO), 2015-2025. Medallion pipeline: bronze, silver, gold.';

CREATE VOLUME IF NOT EXISTS workspace.openalex.raw
COMMENT 'Raw OpenAlex API exports (JSONL), one file per university';

CREATE VOLUME IF NOT EXISTS workspace.openalex.reference
COMMENT 'Indicator tables of the original pandas + SQLite project, used to validate the gold layer';
```
